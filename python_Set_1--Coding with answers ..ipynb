```python
3.Write a Python program to print the sum of numbers from 1 to 10. 
Expected Output:55
```

```python
#Answer:
n=10
sum=0
while n>0:
    sum=sum+n
    n=n-1
print(sum)
```

```python
4.Write a Python function to find the factorial of 5. Expected Output:
120
```

```python
#Answer:
n=5
fac = 1
while n>0:
    fac=fac*n
    n=n-1
print(fac)
```
