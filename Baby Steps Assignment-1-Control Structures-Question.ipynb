```python
bmi=float(input("Enter the BMI Index: "))
print("The BMI Index is ",bmi)
if bmi < 18.5:
    print("Under Weight")
elif 18.5 <= bmi and bmi <= 24.9:
    print("Normal Weight")
elif 25.0 <= bmi and bmi <= 29.9:
    print("Over Weight")
elif 30.0 <= bmi and bmi <= 34.9:
    print("Obese Class 1")
elif 35.0 <= bmi and bmi <= 39.9:
    print("Obese Class 2")
else:
    print("Obese Class 3")
```
