```python
#2.Find the error in the following Python code:
def my_function():
    x += 5
    print(x)

x = 10
my_function()
```

```python
#Ans: x is modified inside the function without being declared as global.

def my_function():
    global x
    x += 5
    print(x)

x = 10
my_function()
```
