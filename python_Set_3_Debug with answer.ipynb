```python
#5.What is wrong with this code?

def is_even(n):
    if n % 2 == 1:
        return True
    else:
        return False
```

```python
#Ans: The even logic is false because the the value divided by 2 and given remainder 0 is even number.
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False
if is_even(5) == True:
    print("Even")
else:
    print("Odd")
```

```python
#9.What will happen when you run the following code?


count = 10

def increment():
    count += 1
    print(count)

increment()
```
